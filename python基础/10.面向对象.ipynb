```python
#类
class Name:
    pass
name1 = Name()
name1.age=10  #动态添加属性 不推荐
name1.id=110
print(name1.__dict__) #字典形式存储对象属性
```

```python
class Car:
    wheel=4 #类属性


    def __init__(self,make,model,year): #第一个必须self 类里面的是方法 类外部是函数
        #`__init__`：构造方法，**创建对象的时候自动调用**。
        #self:实例对象本身
        self.make=make  #实例属性
        self.model=model
        self.year=year
    def run(self):#实例方法
        print("跑",self.make,self.model,self.year)

    def __str__(self): #魔法方法,不用手动调用，合适时机自动调用
        return f"{self.make} {self.model} {self.year}"
    def __eq__(self,other):
        return self.make==other.make



car1=Car(1,"model3",2025)
car2=Car(2,"model4",2025)
print(car1.__dict__)
car1.run()
print(car1==car2) #自动调用了__eq__方法
print(Car.wheel,car1.wheel)
```
