增删改查

```python
d={"zjx":1,"zzz":2,"zda":3,1103:4}
d["asd"]=567 #增
d["zjx"]=666 #改
d.pop("zzz") #删除
del d["zda"]

print(d[1103])
print(d.get("zjx"))
print(d.keys()) #获取所有key
print(d.values()) #获取所有v
print(d.items())  #获取所有k-v
print(d)  #获取所有k-v
```
