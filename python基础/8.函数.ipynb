多个返回值是元组格式

```python
def fun(a,b):
    """
     函数实现了什么功能
    :param a: 参数a
    :param b: 参数b
    :return: 返回元组
    """
    return a,b  #多个返回值是元组格式
print(fun(1,b=2))
A,B = fun(3,4)  #解包
```

global关键字

```python
num = 1 #全局变量
def fun():
    global num  #声明一下要用全局变量的num 可以对全局变量修改
    num += 1
print(num)
fun()
print(num)
```

传参 位置  关键字参数：age=10...位置参数在关键字前面

默认参数（缺省参数）可以为参数设置默认值,默认参数放后面

```python
def fun(a,b=1):
    return a,b
print(fun(1,2))
```

不定长参数  位置在前。关键字传在后

```python
#位置参数
def fun(*a):  #元组类型
    return a
print(fun(1,2,3,4,5,6,7))
```

```python
#关键字传递多个
def fun(**a):  #字典类型
    return a
print(fun(i=0,a=4))
```

参数类型：普通的和函数

```python
def add(x, y):
    return x+y

def calc(x, y, oper):
    return oper(x, y)

result = calc(10, 20, add)
print(result)
```

匿名函数（Lamda表达式）单行表达式

```python
line = lambda :print("_______________")#无参
line()
sum = lambda x,y:x+y  #有参
print(sum(1,2))
```

```python
# 典型
data_list = ["C++", "C", "Python", "Jack", "PHP", "Java", "Go", "JavaScript", "Rust"]
print(data_list)

data_list.sort(key=lambda item : len(item)) # 匿名函数典型的应用场景 .
print(data_list)
```
