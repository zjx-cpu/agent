无法修改 有序 索引（0  -1）[1:5:2] 切片 步长负数表示从后往前截取

```python
s="zjxzjx666"
print(s[1:5:2])
print(s[::2])
print(s[-1])
print(s[::-1])
```

方法

```python
s=" zjxz jx**666 "
print(s.find("z")) #返回第一次位置 没有返回-1
print(s.count("z")) #统计字串在字符串中出现的次数
print(s.upper()) #字母转大写
print(s.lower())  #转小写
print(s.split(' ')) #按照指定分割符分割成列表
print(s.strip()) #去除两端空白字符
print(s.strip('*')) #去除两端指定字符 中间不删
print(s.replace('z','*')) #把z替换为*
print(s.startswith('z')) #是否以指定字串开头
```
