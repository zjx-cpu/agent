切片
[开始：结束：步长]
默认：0开始 末尾结束 步长1
左闭右开  

```python
s=[1,2,3,4,5]
print(s[0:1])
print(s[:1])
```

方法

```python
s=[1,2]
s.append(3) #尾部加
print(s)
s.insert(0,4) #0索引加元素
print(s)
s.remove(1) #移除第一个匹配元素
print(s)
a=s.pop() #删除指定索引 默认删除最后一个,可以返回删除的元素
print(s,"  ",a)
s.sort() #排序
print(s)
s.reverse() #反转
print(s)
```

```python
s=[1,2]
print(sum(s))#sum求和
print(sum(s)/len(s))#len长度
```

合并

```python
a1=[19,25,3,52,85,36]
a2=[42,45,252,3,46,55]
a3=a1+a2  #加号合并
a4=[*a1,*a2]  # *解包，把容器解开，放在 a4 一样
print(a4)
print(a3)
```

列表推导式

```python
a=[i**2 for i in range(1,10)]
print(a)
a=[i**2 for i in range(1,10) if i%2==0]
print(a)
```
