```python
#封装
class Car:
    def __init__(self,name,price,owner):
        self.name=name
        self.price=price
        self.__owner=owner #前面加__表示私有 属性只能在类内部使用
    def get_owner(self):
        return self.__owner
    def __car_owner(self): #__私有方法，前面加__
        print(self.name)
        print(self.price)

if __name__=='__main__':
    car = Car("zjx",300,"zzz")
    print(car.name)
    #print(car.owner) #报错
    print(car.get_owner())#可以用访问私有属性的方法来调用
    print(car._Car__owner) #可以前面加 _类名__属性来调用
    #方法同理
    car._Car__car_owner()
```

```python
#继承，python可以多继承，可以有多个爹
class Car:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def run(self):
        print(self.name,"飞着跑")

class Bike():
    def __init__(self,name,price,year,color):
        self.name=name
        self.price=price
        self.year=year
        self.color=color
    def run(self):
        print(self.name,"自行车着跑")
    def fly(self):
        print("有轮子")

class FuelCar(Car): #Car的子类，继承非私有的方法
    def run(self): #方法重写
        Car.run(self) #也可以调用父类的方法
        super().run() #也可以调用父类的方法
        print("油车跑")

class WenJie(Car,Bike):
    def __init__(self,name,price,year,color):
        super().__init__(self,name,price) #直接调用父类的init方法
        Bike.__init__(self,year,color)

if __name__ == '__main__':
    fuelcar = FuelCar("张嘉兴",300)
    print(fuelcar.__dict__)
    fuelcar.run()

    a=WenJie("h",300,2026,"red")#默认优先使用 第一个类
    print(a.__dict__)
```

```python
#多态  同一个方法有多种不同表现形态
# 定义几个类，都有 speak 方法
class Animal:
    def speak(self):
        print("动物叫")

class Cat(Animal):
    def speak(self):
        print("喵喵喵")

class Bird(Animal):
    def speak(self):
        print("叽叽叽")

# 多态函数：接收任意对象，只要有 speak 方法就能调用
def make_speak(animal:Animal):#类型指定为父类
    animal.speak()

# 测试
cat = Cat()
bird = Bird()

make_speak(cat)
make_speak(bird)
```

```python
# 多态 鸭子类型   Python 不强制要求必须继承父类，只要具备对应的方法即可（鸭子类型）
# 定义几个类，都有 speak 方法
class Dog:
    def speak(self):
        print("汪汪汪")

class Cat:
    def speak(self):
        print("喵喵喵")

class Bird:
    def speak(self):
        print("叽叽叽")

# 多态函数：接收任意对象，只要有 speak 方法就能调用
def make_speak(animal):
    animal.speak()

# 测试
dog = Dog()
cat = Cat()
bird = Bird()

make_speak(dog)
make_speak(cat)
make_speak(bird)
```
