集合 不重复，自动去重，无序，可修改

```python
s1={'1',4,"zjd",4}
s2=set() #空集合
print(s1,type(s1))
```

方法

```python
s1={1,23,4,5,5,2,42}
s2={5,1,6}
s1.add("zjx")
s1.remove(1) #移除指定元素，没有就报错
print(s1.pop()) #随机删除并返回
#s1.clear() #清空集合
print(s1.difference(s2)) #在s1中不在s2中 也可以直接s1-s2
print(s1.union(s2)) #两集合并集 也可以 s1|s2
print(s1.intersection(s2)) #两集合交集 也可以 s1&s2
```
