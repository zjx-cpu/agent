__name__是 Python 内置的**特殊变量**：

```python
if __name__ == "__main__":
    # 只有直接运行这个.py文件的时候，这段代码才执行
    #作为模块运行时，这段代码不执行
    print("直接运行")
```

\_\_all__  当使用 from 模块 import * 时，哪些名字会被导入。

```python
# mymod.py
__all__ = ["add", "subtract"] # 声明 * 只能导入这两个名字

def add(x,y):
    return x+y

def subtract(x,y):
    return x-y

def hello():
    print("hello")

name = "test"
```
