组包（多个值合并一个容器）与解包（解开为独立元素给不同变量）

```python
t = (1,2,2,1)
a,b,c,d = t
A,*B,C=t #*表示收集剩余的元素，生成列表
print(A,B,C)
```

```python
a,b,c=10,20,30
b,a,c=a,b,c  #组包解包实例
print(a,b,c)
```
